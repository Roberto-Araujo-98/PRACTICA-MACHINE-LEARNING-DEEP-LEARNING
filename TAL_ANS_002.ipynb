```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
x = np.array([[100,50], [120,60],[400,200], [380,220], [700,300], [720,320]])
x
```

```python
model = KMeans(n_clusters=3, random_state= 42)
model.fit(x)
```

```python
centros = model.cluster_centers_
etiquetas = model.labels_
```

```python
print("Etiquetas\n", model.labels_)
print("Centroides\n", model.cluster_centers_)
```

```python
plt.scatter(x[:,0], x[:,1], c=etiquetas)
plt.scatter(centros[:,0], centros[:,1], marker = "x", c = "red" )
#
```
