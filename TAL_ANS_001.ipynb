```python
import numpy as np
from sklearn.cluster  import KMeans
```

```python
x = np.array([1,2,2,8,9,2]).reshape(-1,1)
x
```

```python
model = KMeans(n_clusters=2)
```

```python
model.fit(x)
```

```python
model.labels_
```

```python
print("Etiquetas asignadas por KMeans", model.labels_)
print("Centroides asignados por KMeans", model.cluster_centers_)
```

```python
import matplotlib.pyplot as plt
centros = model.cluster_centers_
print(centros)
```

```python
plt.scatter(x.flatten(), model.labels_, c= model.labels_)
```
