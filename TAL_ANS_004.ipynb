```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer()
x =cancer.data
y= cancer.target
```

```python
model = KMeans(n_clusters=2, random_state=42)
model.fit_predict(x) #Entrenamos el modelo (utilizando las 4 caracteristicas)
```

```python
etiquetas = model.labels_
centros = model.cluster_centers_
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,6))
plt.subplot(1,2,1)
plt.title("CLUSTERING POR KMEANS")
plt.scatter(x[:,0],x[:,1], c= etiquetas)
plt.subplot(1,2,2)
plt.title("DATA REAL")
plt.scatter(x[:,0],x[:,1], c=y)
```

```python
print(etiquetas)
print(y)
print("cantidad de clusters", len(set(etiquetas)))
print("muestras por cluster", np.bincount(etiquetas))
print("muestras por caso (real)", np.bincount(y))
```

```python
DIFERENCIA = np.bincount(etiquetas) - np.bincount(y)
DIFERENCIA
```
