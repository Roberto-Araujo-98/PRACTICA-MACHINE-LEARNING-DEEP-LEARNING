```python
import pandas as pd
datos = {
    "experiencia": [1, 2, 3, 5, 7, 10, 4, 6, 8, 2, 9, 3],
    "lenguaje": [
        "python", "java", "csharp", "python",
        "java", "csharp", "java", "python",
        "csharp", "python", "java", "csharp"
    ],
    "rango": [
        "junior", "junior", "intermedio", "intermedio",
        "senior", "senior", "intermedio", "senior",
        "senior", "junior", "senior", "intermedio"
    ],
    "pret_salarial": [
        1500, 1800, 2500, 3000,
        4500, 5000, 2800, 4200,
        4800, 1700, 5500, 2700
    ],
    "contratado": [
        "si", "no", "si", "si",
        "si", "si", "no", "si",
        "si", "no", "si", "no"
    ]
}
df = pd.DataFrame(datos)
print(df)
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder, LabelBinarizer
import numpy as np
```

```python
x_num= df[['experiencia','pret_salarial']]
#x_num
x_nominal= df[['lenguaje']]
#x_nominal
x_ordinal= df[['rango']]
#x_ordinal
```

```python
num_scaler= StandardScaler()
nominal_encoder= OneHotEncoder(sparse_output=False)
ordinal_encoder= LabelEncoder()
```

```python
x_num_scaled= num_scaler.fit_transform(x_num)
#x_num_scaled
x_nominal_encoded= nominal_encoder.fit_transform(x_nominal)
#x_nominal_encoded
x_ordinal_encoded= ordinal_encoder.fit_transform(x_ordinal)
#x_ordinal_encoded
```

```python
y_cat= df[['contratado']]
#y_cat
```

```python
label_binarizer= LabelBinarizer() #BINARIZAR ETIQUETAS (Y)
y_cat_binarized= label_binarizer.fit_transform(y_cat)
#y_cat_binarized
```

```python
np.hstack([x_num_scaled, x_nominal_encoded, x_ordinal_encoded.reshape(-1,1), y_cat_binarized])
x= np.hstack([x_num_scaled, x_nominal_encoded, x_ordinal_encoded.reshape(-1,1)])
y= y_cat_binarized.reshape(-1)
print(x.shape)
print(y.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier
model= DecisionTreeClassifier()
model.fit(x,y)
y_pred=model.predict(x)
print(y_pred)
print(y)
#IMPLEMENTAR TU RNA PARA ENTRENAR Y CLASIFICAR CON TODOS LOS DATOS DEL DS
```

```python
import tensorflow as tf
from sklearn.metrics import accuracy_score, classification_report
```

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.Input(shape=(6,)),
    tf.keras.layers.Dense(16, activation='relu', input_shape=(2,)),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.fit(x, y, epochs=100, batch_size=4)
```

```python
y_pred = model.predict(x)
```

```python
print(accuracy_score(y, y_pred.round()))
print(classification_report(y, y_pred.round()))
```
