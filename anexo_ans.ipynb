```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

X = StandardScaler().fit_transform(load_iris().data)
```

```python
#CELDA 1: Inercia para k = 1 hasta 10
ks = list(range(1, 11))
inercias = []
for k in ks:
    modelo = KMeans(n_clusters=k, n_init=10, random_state=42)
    modelo.fit(X)
    inercias.append(modelo.inertia_)
```

```python
#CELDA 2: Distancia de cada punto a la recta que une el primero con el último
x1, y1 = ks[0], inercias[0]
x2, y2 = ks[-1], inercias[-1]

distancias = []
for k, inercia in zip(ks, inercias):
    d = abs((y2 - y1) * k - (x2 - x1) * inercia + x2 * y1 - y2 * x1)
    distancias.append(d)

#CELDA 3: El codo es el punto más alejado de esa recta
k_optimo = ks[np.argmax(distancias)]
print("Número óptimo de clusters:", k_optimo)
```
