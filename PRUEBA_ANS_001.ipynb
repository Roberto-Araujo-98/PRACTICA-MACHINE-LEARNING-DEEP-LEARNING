```python
import numpy as np
from sklearn.cluster import KMeans

x = np.array([2,4,5,20,21,25,50,40]).reshape(-1,1)
model = KMeans(n_clusters= 3)
model.fit(x)

print("Etiquetas", model.labels_)
print("Centroides", model.cluster_centers_)
```

```python
import matplotlib.pyplot as plt
```

## Gráfico KMeans (n_clusters= 3)


```python
plt.scatter( x.flatten(),model.labels_, c = model.labels_)
```
