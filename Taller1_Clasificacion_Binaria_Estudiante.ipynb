# Taller 1 (Estudiante): Clasificacion Binaria
## Caso: ¿El estudiante aprueba o reprueba el curso?

### Tu mision:
Debes armar y calibrar una red neuronal que aprenda a predecir si un estudiante aprueba o no.

### Lo que SI puedes modificar para mejorar el modelo:
* **Neuronas por capa**: Puedes cambiar de 8 a 16, 32, etc.
* **Capas ocultas**: Puedes agregar o quitar capas intermedias `Dense(..., activation='relu')`.
* **Épocas**: Puedes probar con 30, 60 u 80 pasadas.
* **Batch size**: Tamaño de grupos de datos (ejemplo 8, 16 o 32).

### Lo que NO debes cambiar (Reglas matematicas):
* La capa de salida DEBE tener `1` neurona.
* La activación final DEBE ser `'sigmoid'` (porque da la probabilidad de aprobación).
* La pérdida (loss) DEBE ser `'binary_crossentropy'`.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder
from sklearn.compose import ColumnTransformer
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Datos
n = 400
horas_estudio = np.random.uniform(2, 20, n)
asistencia = np.random.uniform(55, 100, n)
tareas = np.random.choice(['si', 'no'], size=n, p=[0.75, 0.25])
puntos = (horas_estudio * 0.4) + (asistencia * 0.08) + np.where(tareas == 'si', 3.0, 0.0)
estado = np.where(puntos >= 12.0, 'Aprobado', 'Reprobado')

df = pd.DataFrame({
    'horas_estudio_semanal': np.round(horas_estudio, 1),
    'asistencia_porcentaje': np.round(asistencia, 1),
    'hizo_tareas': tareas,
    'estado': estado
})

le = LabelEncoder()
y = le.fit_transform(df['estado'])

X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['estado']), y, test_size=0.2, random_state=42, stratify=y)

pre = ColumnTransformer([
    ('num', StandardScaler(), ['horas_estudio_semanal', 'asistencia_porcentaje']),
    ('cat', OneHotEncoder(drop='first', sparse_output=False), ['hizo_tareas'])
])

X_train_proc = pre.fit_transform(X_train)
X_test_proc = pre.transform(X_test)
print("Datos listos. Filas de entrenamiento:", len(X_train_proc))
```

### Tu turno: Ajusta la red neuronal y entrena


```python
# MODIFICA AQUI LOS HIPERPARAMETROS
modelo_estudiante = Sequential([
    Dense(units=8, activation='relu', input_shape=(X_train_proc.shape[1],)),
    # Puedes agregar otra capa intermedia aqui si quieres:
    Dense(units=4, activation='relu'),
    # Capa de salida obligatoria:
    Dense(units=1, activation='sigmoid')
])

modelo_estudiante.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Entrena:
historia = modelo_estudiante.fit(
    X_train_proc,
    y_train,
    epochs=40,        # Modifica la cantidad de epocas
    batch_size=16,    # Modifica el batch size
    validation_split=0.2,
    verbose=1
)

loss, acc = modelo_estudiante.evaluate(X_test_proc, y_test, verbose=0)
print(f"\n--> Tu porcentaje de acierto alcanzado: {acc * 100:.1f} %")
# Meta: Lograr mas del 88% de precision
```
