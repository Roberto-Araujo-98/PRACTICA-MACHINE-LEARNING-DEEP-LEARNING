```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler #Normalizar la data
from sklearn.datasets import load_iris #Dataset
from sklearn.decomposition import PCA #Reducción de dimensiones
from sklearn.cluster import KMeans #Algoritmo de clustering
from sklearn.metrics import silhouette_score, adjusted_rand_score, normalized_mutual_info_score #Metricas para evaluar
```

```python
iris= load_iris()
y = iris.target #Obtenemos datos para entrenar el modelo
x = iris.data #Etiquetas para las métricas (El modelo no requiere para su entrenamiento)
nombres = iris.target_names
print(nombres)
```

```python
x_scaled = StandardScaler().fit_transform(x) # Escalamos los datos obtenidos (para el entrenamiento)
x_scaled
```

```python
x_2d = PCA(n_components=2).fit_transform(x_scaled) #Reducimos a 2d las 4 columnas (Para graficar los datos)
x_2d
```

```python
model = KMeans(n_clusters= 3, random_state= 42, n_init = 10)
model.fit(x_scaled)
```

```python
# GRAFICANDO DATOS CON LA PCA APLICADA SOBRE LA DATA ESCALADA ANTERIORMENTE
plt.figure(figsize=(10,6))
plt.subplot(1,2,1)
plt.scatter(x_2d[:,0], x_2d[:,1], c=model.labels_)

plt.subplot(1,2,2)
plt.title('Datos Reales PCA')
plt.scatter(x_2d[:,0], x_2d[:,1], c=y)
plt.show()
```

```python
print("Metricas del modelo KMeans")
print(silhouette_score(x_scaled, model.labels_)) #La silueta no requiere las etiquetas reales
print(adjusted_rand_score(y, model.labels_)) #Score rand ajustado si requiere las etiquetas reales
print(normalized_mutual_info_score(y, model.labels_)) #Datos compartidos normalizados si requiere las etiquetas reales
#Interpretación: 0 = ALEATORIO, , menor a 0 = Peor que aleatorio, cerca a 1 = Métrica aceptable mejor a clustering aleatorio
```
