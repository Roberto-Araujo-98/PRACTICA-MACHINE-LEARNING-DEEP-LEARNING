## Recomendador de Bebidas (3 Clases)

### Tu mision:
Crear una red neuronal que recomiende una de las 3 bebidas posibles ('Agua_Jugo', 'Cafe' o 'Batido').

### Lo que SI puedes modificar para experimentar:
* Cantidad de capas intermedias y cantidad de neuronas (ej. 16, 32, 64).
* Cantidad de épocas de entrenamiento (ej. 40, 60, 100).
* Tamaño de batch (8, 16, 32).

### Lo que NO debes cambiar (Reglas estrictas de Multiclase):
* La capa de salida DEBE tener `3` neuronas (una para cada bebida).
* La activación final DEBE ser `'softmax'` (distribuye la probabilidad sumando 100%).
* La pérdida (loss) DEBE ser `'sparse_categorical_crossentropy'`.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder
from sklearn.compose import ColumnTransformer
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Datos
n = 600
edad = np.random.randint(16, 65, n)
cansancio = np.random.randint(1, 11, n)
ejercicio = np.random.choice(['si', 'no'], size=n, p=[0.4, 0.6])

bebida = []
for e, c, ej in zip(edad, cansancio, ejercicio):
    if ej == 'si':
        bebida.append('Batido')
    elif c >= 6:
        bebida.append('Cafe')
    else:
        bebida.append('Agua_Jugo')

df = pd.DataFrame({'edad': edad, 'nivel_cansancio': cansancio, 'hizo_ejercicio_hoy': ejercicio, 'tipo_bebida': bebida})

le = LabelEncoder()
y = le.fit_transform(df['tipo_bebida'])

X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['tipo_bebida']), y, test_size=0.2, random_state=42, stratify=y)

pre = ColumnTransformer([
    ('num', StandardScaler(), ['edad', 'nivel_cansancio']),
    ('cat', OneHotEncoder(drop='first', sparse_output=False), ['hizo_ejercicio_hoy'])
])

X_train_proc = pre.fit_transform(X_train)
X_test_proc = pre.transform(X_test)
print("Datos preparados. Clases a predecir:", le.classes_)
```

### Tu turno: Modifica la arquitectura de la red


```python
# COMPLETA Y AJUSTA LA RED:
modelo_estudiante = Sequential([
    Dense(units=16, activation='relu', input_shape=(X_train_proc.shape[1],)),
    # Puedes modificar o agregar capas intermedias:
    Dense(units=8, activation='relu'),
    # Capa de salida OBLIGATORIA (3 clases con softmax):
    Dense(units=3, activation='softmax')
])

modelo_estudiante.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Entrena:
historia = modelo_estudiante.fit(
    X_train_proc,
    y_train,
    epochs=40,        # Ajusta las epocas
    batch_size=16,    # Ajusta el batch size
    validation_split=0.2,
    verbose=1
)

loss, acc = modelo_estudiante.evaluate(X_test_proc, y_test, verbose=0)
print(f"\n--> Exactitud alcanzada en tu recomendador: {acc * 100:.1f} %")
# Meta: Superar el 90% de precision
```
